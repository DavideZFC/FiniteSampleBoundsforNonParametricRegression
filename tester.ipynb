```python
import numpy as np
vector = np.array([0, 1.0, 3, 0, 2]).astype(int)

def from_numbers_to_bins(vector):
    ans = np.zeros(np.sum(vector))
    curr = 0
    for i,num in enumerate(vector):
        for j in range(num):
            ans[curr] = i
            curr += 1
    return ans


print(from_numbers_to_bins(vector))
```
